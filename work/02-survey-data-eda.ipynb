Examining Demogaphic Variables

```python
# bar charts
survey['Q-demos-age'].value_counts().plot(kind='bar')
# 25-34 is the most common age group, while 65 and older is the least common
# Important to note that 18-24 accounts for a smaller range of ages, which may contribute to it appearing less frequently in the data
```

```python
survey['Q-demos-hispanic'].value_counts().plot(kind='bar')
# mostly non-hispanic participants
```

```python
survey['Q-demos-race'].value_counts() # too many unique values for bar chart, printing value counts instead
# It appears that this question allowed for multiple selections
# Rows with multiple responses should be separated into multiple rows, one for each selection
```

```python
survey['Q-demos-education'].value_counts().plot(kind='bar')
# bachelor's degree is most common, followed by high school diploma
# Very few participants have less than a high school dploma or did not disclose
```

```python
survey['Q-demos-income'].value_counts().plot(kind='bar')
# Most common income range is 25,000-49,999, followed closely by 50,000-75,000
```

```python
survey['Q-demos-gender'].value_counts().plot(kind='bar')
# More females than mals in the dataset, although they are relatively evenly split
```

```python
survey['Q-sexual-orientation'].value_counts().plot(kind='bar')
# Majority of participants identify as straight
```

```python
survey['Q-demos-state'].value_counts() # too many unique values for bar chart, printing value counts instead
# California, Texas, Florida, and New York are the most common states in the dataset
# This coul be related to state populations
```

Examining Amazon Usage Related Variables

```python
survey['Q-amazon-use-howmany'].value_counts().plot(kind='bar')
# majority of participants are the sole users of their amazon accounts
```

```python
survey['Q-amazon-use-hh-size'].value_counts().plot(kind='bar')
# despite the majority of participants being the only users of their amazon accounts, the most common household size of survey participants is 2, followed by 4+
```

```python
survey['Q-amazon-use-how-oft'].value_counts().plot(kind='bar')
# majority of participants use their amazon accounts less than 5 times per month
```
